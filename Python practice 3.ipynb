```python
# Q6. Simple Calculator Function
def addition(a, b):
    return a + b
def subtraction(a, b):
    return a - b
def multiplication(a, b):
    return a * b
def division(a, b):
    if b != 0:
        return a / b
    return "Division by zero not allowed"

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

print("Addition:", addition(a, b))
print("Subtraction:", subtraction(a, b))
print("Multiplication:", multiplication(a, b))
print("Division:", division(a, b))
```

```python
# Q9. Factorial using Function
def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact

num = int(input("Enter number: "))
print("Factorial:", factorial(num))
```

```python
# Q10. Prime Number Checker
def is_prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

num = int(input("Enter number: "))

if is_prime(num):
    print("Prime Number")
else:
    print("Not Prime")
```

```python
# Q11. Attendance System
students = []

def add_student(name):
    students.append(name)

def remove_student(name):
    if name in students:
        students.remove(name)

def display_students():
    print("Students List:")
    for s in students:
        print(s)

add_student("Rahul")
add_student("Anita")
remove_student("Rahul")

display_students()
```

```python
# Q12. Employee Salary Dictionary
employees = {}

def add_employee(name, salary):
    employees[name] = salary

def update_salary(name, salary):
    if name in employees:
        employees[name] = salary

def search_employee(name):
    if name in employees:
        print(name, "Salary:", employees[name])
    else:
        print("Employee not found")

add_employee("Amit", 50000)
update_salary("Amit", 55000)
search_employee("Amit")
```

```python
# Q15. Word Frequency Counter
sentence = input("Enter sentence: ")
words = sentence.split()
freq = {}
for word in words:
    if word in freq:
        freq[word] += 1
    else:
        freq[word] = 1

print(freq)
```

```python
# Q16. Library Book System
books = []
def add_book(book):
    books.append(book)
def remove_book(book):
    if book in books:
        books.remove(book)
def search_book(book):
    if book in books:
        print("Book Found")
    else:
        print("Book Not Found")
def display_books():
    print("Books List:")
    for b in books:
        print(b)
while True:
    print("\n1.Add Book")
    print("2.Remove Book")
    print("3.Search Book")
    print("4.Display Books")
    print("5.Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        book = input("Enter book name: ")
        add_book(book)

    elif choice == 2:
        book = input("Enter book name: ")
        remove_book(book)

    elif choice == 3:
        book = input("Enter book name: ")
        search_book(book)

    elif choice == 4:
        display_books()

    elif choice == 5:
        break
```

```python
# Q17. Student Result Analyzer
marks = []
for i in range(5):
    m = int(input("Enter marks: "))
    marks.append(m)
def average(marks):
    return sum(marks) / len(marks)
def highest(marks):
    return max(marks)
def lowest(marks):
    return min(marks)
def passed_students(marks):
    count = 0
    for m in marks:
        if m > 40:
            count += 1
    return count
print("Average:", average(marks))
print("Highest:", highest(marks))
print("Lowest:", lowest(marks))
print("Passed Students:", passed_students(marks))
```

```python
# Q18. Password Strength Checker
def check_password(password):

    if (len(password) >= 8 and
        any(ch.isdigit() for ch in password) and
        any(ch.isupper() for ch in password) and
        any(ch.islower() for ch in password)):

        return "Strong Password"

    return "Weak Password"

pwd = input("Enter password: ")
print(check_password(pwd))
```

```python
# Q19. Contact Book Application
contacts = {}
while True:
    print("\n1.Add Contact")
    print("2.Search Contact")
    print("3.Delete Contact")
    print("4.Display Contacts")
    print("5.Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:
        name = input("Enter name: ")
        phone = input("Enter phone: ")
        contacts[name] = phone

    elif choice == 2:
        name = input("Enter name: ")

        if name in contacts:
            print("Phone:", contacts[name])
        else:
            print("Contact not found")

    elif choice == 3:
        name = input("Enter name: ")

        if name in contacts:
            del contacts[name]

    elif choice == 4:
        print(contacts)

    elif choice == 5:
        break
```

```python
# Q20. Mini Banking System
balance = 1000
def deposit(amount):
    global balance
    balance += amount
    print("Deposited:", amount)
def withdraw(amount):
    global balance

    if amount <= balance:
        balance -= amount
        print("Withdrawn:", amount)
    else:
        print("Insufficient Balance")
def check_balance():
    print("Current Balance:", balance)

deposit(500)
withdraw(300)
check_balance()
```

```python
# Q21. HackerRank Style Problem
numbers = [10, 20, 30, 40, 50]
# Find maximum
max_num = numbers[0]
for num in numbers:
    if num > max_num:
        max_num = num
# Find minimum
min_num = numbers[0]
for num in numbers:
    if num < min_num:
        min_num = num
# Reverse list
reversed_list = []
for i in range(len(numbers)-1, -1, -1):
    reversed_list.append(numbers[i])

print("Maximum:", max_num)
print("Minimum:", min_num)
print("Reversed List:", reversed_list)
```
